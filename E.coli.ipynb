```python
Una cepa de E.coli tiene una tasa de crecimiento tal que un individuo se duplica
cada 30 minutos
Calcula en cuento tiempo esta cepa:
a) LLenaria el  Volumen de la Tierra (considerando como una esfera nueva)
b) Cubriria la superficie de la Tierra
c) Repete el calculo para una tasa de crecimiento:
     i) Se duplica cada hora
    ii) se duplica cada 1/4 hora
```

```python
import math
import pandas as pd
import math as mt
import numpy as np
import matplotlib.pyplot as plt
```

```python
rT = 6371
vT = (4/3)*np.pi*(rT**3)
sT = 4*np.pi*(rT**2)
print( f"volumen de la tierra: {vT} m³")
print (f"superficie de la tierra {sT} m²")
```

```python
rEC = 0.5e-6
vEC = (4/3)*np.pi*(rEC**3)
sEC = np.pi * (rEC ** 2)
N_volumen = (vT*(10**3))/(vEC*(10**-6))
print(f"Se necesitan {N_volumen} e.coli para llenar el volumen de la Tierra")
```

```python
#Relleno del volumen de la Tierra cada 60 minutos
t = math.log2(N_volumen)
t_dias = t/24
print (f"Se necesitaran {t} horas o {t_dias} días para rellenar la tierra de E. coli si se duplican cada 60m")
```

```python
#Relleno del volumen de la Tierra cada 30 minutos
t_30m = t/2
t_30m_dias = (t_30m/24)
print (f"Se necesitaran {t_30m} horas o {t_30m_dias} días para rellenar la tierra de E. coli si se duplican cada 30m")
```

```python
#Relleno del volumen de la Tierra cada 15 minutos
t_15m = t/4
t_15m_dias = t_15m/24
print (f"Se necesitaran {t_15m} horas o {t_15m_dias} días para rellenar la tierra de E. coli si se duplican cada 15m")
```

```python
#Llenado de la superficie de la Tierra cada 60 minutos
N_superficie = sT/(sEC*(10**-6))
t_superficie = math.log2(N_superficie)
t_superficie_dias = t_superficie/24
print (f"Se necesitaran {t_superficie} horas o {t_superficie_dias} días para llenar la superficie de la tierra de E. coli si se duplican cada 60m")
```

```python
#Llenado de la superficie de la Tierra cada 30 minutos
t_superficie30m = t_superficie/2
t_superficie30m_dias = t_superficie30m/24
print (f"Se necesitaran {t_superficie30m} horas o {t_superficie30m_dias} días para llenar la superficie de la tierra de E. coli si se duplican cada 30m")
```

```python
#Llenado de la superficie de la Tierra cada 15 minutos
t_superficie15m = t_superficie/4
t_superficie15m_dias = t_superficie15m/24
print (f"Se necesitaran {t_superficie15m} horas o {t_superficie15m_dias} días para llenar la superficie de la tierra de E. coli si se duplican cada 15m")
```
